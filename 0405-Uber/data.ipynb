```python
from ucimlrepo import fetch_ucirepo
import pandas as pd
from pathlib import Path

# 获取数据
us_census_data_1990 = fetch_ucirepo(id=116)

df = us_census_data_1990.data.original
variables = us_census_data_1990.variables
# metadata = us_census_data_1990.metadata

# 输出目录
out_dir = Path("us_census_1990_inspect")
out_dir.mkdir(exist_ok=True)

# 1) 所有列名
with open(out_dir / "all_columns.txt", "w", encoding="utf-8") as f:
    f.write(f"Total columns: {len(df.columns)}\n\n")
    for i, col in enumerate(df.columns, 1):
        f.write(f"{i:02d}. {col}\n")

# 2) 前5行完整内容（取消列省略）
with pd.option_context(
    "display.max_columns", None,
    "display.max_colwidth", None,
    "display.width", 100000
):
    full_head_str = df.head().to_string(index=True)

with open(out_dir / "head_5_full.txt", "w", encoding="utf-8") as f:
    f.write(full_head_str)

# 3) 前5行转置，便于逐列查看
with pd.option_context(
    "display.max_rows", None,
    "display.max_colwidth", None,
    "display.width", 100000
):
    head_transposed_str = df.head().T.to_string()

with open(out_dir / "head_5_transposed.txt", "w", encoding="utf-8") as f:
    f.write(head_transposed_str)

# 4) 保存前5行为 csv
df.head().to_csv(out_dir / "head_5_full.csv", index=False)

# 5) 保存全部变量说明
if variables is not None:
    variables.to_csv(out_dir / "variables.csv", index=False)

# # 6) 保存元数据
# with open(out_dir / "metadata.txt", "w", encoding="utf-8") as f:
#     f.write(str(metadata))

print("已输出到目录：", out_dir.resolve())
print("生成文件：")
for p in sorted(out_dir.iterdir()):
    print("-", p.name)
```
